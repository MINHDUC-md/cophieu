```python
# Import các thư viện cần thiết
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
from transformers import BertTokenizer, BertForSequenceClassification
import torch
import nltk
import ipywidgets as widgets
from IPython.display import display, HTML

# Tạo quyền truy cập với Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Đọc dữ liệu từ tệp CSV (Tải bộ dữ liệu IMDB hoặc thay bằng bộ dữ liệu khác)
data = pd.read_csv('/content/drive/MyDrive/MyDrive/IMDB Dataset.csv/IMDB Dataset.csv')

# Tiền xử lý dữ liệu
nltk.download('stopwords')
from nltk.corpus import stopwords

def preprocess_text(text):
    stop_words = set(stopwords.words('english'))
    # Xóa ký tự không cần thiết, chuyển tất cả thành chữ thường, loại bỏ stopwords
    text = text.lower()
    text = ''.join(e for e in text if e.isalnum() or e.isspace())
    text = ' '.join([word for word in text.split() if word not in stop_words])
    return text

# Tải mô hình BERT
model = BertForSequenceClassification.from_pretrained('distilbert-base-uncased', num_labels=2)
tokenizer = BertTokenizer.from_pretrained('distilbert-base-uncased')
model.save_pretrained("./bert_model")
tokenizer.save_pretrained("./bert_model")
model = BertForSequenceClassification.from_pretrained("./bert_model")
tokenizer = BertTokenizer.from_pretrained("./bert_model")


# Hàm dự đoán cảm xúc của câu nói
def predict_sentiment(text):
    encodings = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=512)
    with torch.no_grad():
        outputs = model(**encodings)
        logits = outputs.logits
    prediction = torch.argmax(logits, dim=1).item()
    return 'positive' if prediction == 1 else 'negative'

# Tiền xử lý và biểu diễn văn bản bằng TF-IDF
data['review'] = data['review'].apply(preprocess_text)
X = data['review']
y = data['sentiment']

# Biểu diễn văn bản bằng TF-IDF
vectorizer = TfidfVectorizer(max_features=5000)
X_tfidf = vectorizer.fit_transform(X)

# Chia dữ liệu thành tập huấn luyện và kiểm tra
X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y, test_size=0.2, random_state=42)

# Huấn luyện mô hình Naive Bayes
model_naive_bayes = MultinomialNB()
model_naive_bayes.fit(X_train, y_train)

# Dự đoán trên tập kiểm tra và đánh giá mô hình Naive Bayes
y_pred = model_naive_bayes.predict(X_test)
print(f'Naive Bayes Accuracy: {accuracy_score(y_test, y_pred)}')
print(f'Classification Report:\n{classification_report(y_test, y_pred)}')

# Tạo giao diện với background và label

# HTML để tạo background cho ứng dụng
HTML("""
<style>
    .container {
        background-color: #f0f0f0;
        padding: 20px;
        border-radius: 8px;
        width: 60%;
        margin: auto;
        font-family: Arial, sans-serif;
    }
    .header {
        text-align: center;
        font-size: 30px;
        color: #4CAF50;
    }
    .label {
        font-size: 18px;
        font-weight: bold;
    }
    .output {
        background-color: #ffffff;
        padding: 10px;
        border: 1px solid #ddd;
        border-radius: 5px;
        margin-top: 20px;
    }
</style>
""")

# Tạo các widget
text_input = widgets.Text(
    value='',
    placeholder='Type a sentence for sentiment analysis...',
    description='Input:',
    disabled=False
)

output = widgets.Output()

# Hàm để xử lý sự kiện khi người dùng nhấn Enter
def on_text_submit(text_input):
    with output:
        # Nếu có nhập văn bản, thì thực hiện phân tích cảm xúc
        if text_input.value.strip() != '':  # Đảm bảo văn bản không rỗng
            sentiment = predict_sentiment(text_input.value)
            print(f"The sentiment of the input text is: {sentiment}")

# Thiết lập sự kiện khi người dùng nhấn Enter
text_input.on_submit(on_text_submit)

# Tạo giao diện
container = widgets.VBox([
    widgets.HTML('<div class="header">Sentiment Analysis Application</div>'),
    widgets.HTML('<div class="label">Enter text for sentiment analysis:</div>'),
    text_input,
    widgets.HTML('<div class="label">Analysis Result:</div>'),
    output
])

# Hiển thị giao diện
display(container)
```
